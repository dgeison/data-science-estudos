# Métodos de interpolação

```python
import pandas as pd
```

```python
data = [0.5, None,None, 0.52,0.54,None,None,0.59,0.6,None, 0.7 ]
```

```python
s = pd.Series(data)
s
```

```python
s.fillna(0)
```

```python
s.fillna(method= 'ffill')
```

```python
s.fillna(method= 'bfill')
```

```python
s.fillna(s.mean())
```

```python
s.fillna(method= 'ffill', limit=1)
```

```python
s.fillna(method= 'bfill', limit=1)
```
